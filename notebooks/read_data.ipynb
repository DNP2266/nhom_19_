```python
data = data.rename(columns={'Year_y': 'Year'})
data = data.drop(columns=['Year_x'])
```

```python
print(data.dtypes)
```

```python
import time
```

```python
data['Time'] = pd.to_datetime(data['Time'], format='%H:%M')
data['Date'] = pd.to_datetime(data['Date'], format='%Y-%m-%d')
data['Year'] = pd.to_datetime(data['Year'], format = '%Y')
print(data.dtypes)
```

```python
col = data.columns.to_list()
```

```python
for c in col:
    if data[c].value_counts().min() > 1 and len(data[c].value_counts())<100 :
        print(c)
        print(c,data[c].value_counts())
    print('-'*20)
```
